```python
# RUN ME FIRST TO IMPORT EVERYTHING!!
import numpy as np
import matplotlib.pyplot as plt
import math
import scipy.constants as konst
# Now make sure that each of the support methods are recognized. In many instances I've had a markdown associated with the definition of the function. 
```

```python
def Conduction(K, l, A, dT):
    """
    @brief Simple conduction through a extrusion
    K - thermal conductivity in W/(m-K)
    return Q-dot the heat flow through an element of:
    @param l - length (m)
    @param A - cross sectional area (m^2)
    @param dT - difference in temperature along the two ends (K)
    This is a simple case. In general it should obey the heat equation. 
    """

    Qdot = K * A/l * dT
    return Qdot
```

<a href = https://en.wikipedia.org/wiki/Heat_transfer_coefficient> Heat Transfer Coefficient </a> <br> 
$ h = {q \over {\Delta T}} $ where q is the heat flux in ${W\over {m^2}}, q = {{d\dot{Q}}\over{dA}}$ 

```python
def Convection(h, A, T, Tf):
    """
    Convection of a gas, https://en.wikipedia.org/wiki/Convection_(heat_transfer)
    Return the heat transfered per unit time = Watts
    @param h - heat transfer coefficient
    @param A - area in m^2
    @param T - temperature of surface (K)
    @param Tf - temperature of fluid (K)
    """
    Q_dot = h * A * (T-Tf)
    return Q_dot
```

Radiative heat transfer 
<hr>
<a href = https://courses.lumenlearning.com/suny-physics/chapter/14-7-radiation/#:~:text=The%20rate%20of%20heat%20transfer%20by%20emitted%20radiation%20is%20determined,its%20absolute%20temperature%20in%20kelvin. > Radiative Transfer </a> <br>
Stefan-Boltzman law ${Q\over{t}} = \sigma e A T^4$ where $\sigma = 5.67 10^{-8} {J\over s}{m^2}{K^4}$ and e is the emissivity of the object. <br> 
In more general terms: ${Q_{net}\over{t}} = \sigma e A ({T_2}^4 - {T_1}^4)$

```python
def Radative(E, A, T1, T2):
    """
    This in general is very gross. 
    return the heat transfered in Watts
    @param E - emissivity (unitless)
    @param A - area (m^2)
    @param T1 temperature of surface in K
    """
    Qdot = konst.sigma * E * A * (np.power(T2,4) - np.power(T1,4))
    return Qdot
```

This is a specialized form for long concentric cylinders. <br> 
${{A_1}\over{A_2}} = {{r_1}\over{r_2}}$ <br> 
$F_{12} = 1$ <br> 
$\dot{Q} = {{\sigma A_{1}({T_1}^4 - {T_2}^4)}\over{{1\over{\epsilon_1}}}+ {{1-\epsilon_2}\over{\epsilon_2}}({r_1 \over{r_2}})^2}$

```python
def RadiativeCylinder (r1, r2, E1, E2, z, T1, T2):
    """
    @brief - refined version of Cylinder 1/2 radius heat transfered 
    More refined version for a cylinder
    return the heat transfered in Watts
    @param r1 inner radius
    @param r2 outer radius
    @param E1 - emissivity (unitless)
    @param E2 - emissivity (unitless)
    @param z  - length of tubes (calculate Area)
    @param T1 temperature of surface in K
    @param T2 temperature of surface in K
    """
    A1 = 2.0 * np.pi * r1 * z
    A2 = 2.0 * np.pi * r2 * z

    Qdot = konst.sigma * A1 * (np.power(T1,4) - np.power(T2,4))/(1/E1 + (1-E2)/E2 * np.power(r1/r2,2))
    return Qdot
```

<hr>
<a href = https://en.wikipedia.org/wiki/Planck%27s_law> Plank's Radiation Law </a> <br> 
${u_{\nu}} = {{8 \pi h \nu^3}\over{c^3}} {1 \over{e^{{h \nu}\over{k_{B} T}}-1}}$ <br>
describes the spectral emissive power per unit area, per unit solid angle and per unit frequency for particular radiation frequencies. <br> 
I was curious to see how much incident microwave radiation was on the coax. Turns out not much. 
<hr>

```python
def PlankBlackBody(Temperature, Frequency):
    """
    Reference: https://en.wikipedia.org/wiki/Planck%27s_law
    @param Temperature in Kelvin
    @param Frequency in Hz
    """
    Power = 8.0 * np.pi * konst.h * np.power(Temperature, 3.0)/np.power(konst.c,3.0)/(np.exp(konst.h*Frequency/konst.k/Temperature) - 1.0)
    return Power
```

```python
def CylinderArea(radius, length):
    """
    @brief - compute cylinder surface area. 
    determine the area of a cylinder in mks given the radius and length. 
    @param radius - in m
    @param length in m
    """
    Circumference = 2.0 * np.pi *radius
    return (Circumference * length)
```

```python
"""
@brief Start to define the actual device. 
Make shields down to 800mK, but base it on the packing of the coaxial lines. 
"""
class QTunnel(object):
    def __init__(self):
        """
        @brief - define all the parameters of the tunnel. 
        All internal measuerements are in MKS.
        if a k preceeds a variable, it means that it should be a constant
        """
        self.debug = False
        # define coax parameters
        self.__kCoaxDiameter= 0.0022
        self.__kNumberCoax  = 16
        self.__kStackCoax   = 2

        # define temperature stages
        self.__kT10mk       = 0.010    # Kelvin
        self.__kT800mk      = 0.800
        self.__kT4k         = 4.0
        self.__kT50k        = 50.0
        self.__kT300k       = 300.0

        # define the shields of the tunnel. 
        self.__kRStep                = 0.005 
        self.__kInnerShieldThickness = 0.0005 # Gold plated copper. 1.6u Gold Plating should have 0.01-0.1 emmisivity. 
        self.__k50kShieldThickness   = 0.005  # polished Aluminum   should have 0.02-0.2 emissivity. 
        self.__kVacuumShield         = 0.01   # polished 310 SS


        
        # define the Shields in the Cryostat, inner radius. 
        self.__kVacuum_r            = 0.680/2.0  # radius of outer vacuum vessel in m
        self.__k50_r                = 0.621/2.0  # 50k shield
        self.__k4_r                 = 0.551/2.0  # 4k shield
        self.__k800_r               = 0.520/2.0  # 800mk Shield, not used.
        self.__k10_r                = 0.500/2.0  # 10mk Cold plate

        # define tunnel length
        self.__kCoaxLength          = 1.0    # meters
        # 
        # lets get a little more realistic about this, take away from the coax length 
        # the distance inside the cryostat that it has to traverse. 
        # add in an additional 2 cm per side for connections. 
        #
        # This length is the tunnel length. or separation between cryos. 
        self.__kLength              =  self.__kCoaxLength - (self.__kVacuum_r - self.__k10_r)*2.0 - 0.02
        #print('DEBUG coax: ', self.__kCoaxLength , " Radial difference: ",  2*(self.__kVacuum_r - self.__k10_r))

        self.__kE_Gold              = 0.05       # kind of an in-between choice for gold. 
        self.__kE_Al                = 0.01
        self.__kK_Insulator         = 8.0e-7  # PEEK
        self.__kK_InsulatorName     = 'PEEK'

        """
        Derived parameters
        Tunnel Dimensions meters. This is parametric and based on the above.
        This is the inner dimension not the thickness of the material.
        These are all diameters, hence fD. f - indicates class parameter
        """
        self.__fD_Inner  = 0.0       # inner most diameter, coax support
        self.__fD_100mk  = 0.0
        self.__fD_800mk  = 0.0
        self.__fD_4k     = 0.0
        self.__fD_50k    = 0.0
        self.__fD_300k   = 0.0
        # fill these parameters up. 
        self.CalculateTunnel()

        # Store QDot per stage here, index: 0 - Radiation, 1 - Conductive
        self.__QDot_Inner = np.zeros(2)
        self.__QDot_800mk = np.zeros(2)
        self.__QDot_4k    = np.zeros(2)
        self.__QDot_50k   = np.zeros(2)
        self.CalculateQDot()
        
    def Round(self, x):
        """
        @brief - helper function to round up the radii in Calculate tunnel. 
        Want to round up to nearest cm.
        """
        return np.ceil(x*100.0)/100.0
        
    def CalculateTunnel(self):
        """
        @brief - given the coaxial parameters, define the tunnel. 
        """
        self.__fCoax_X   = self.Round(self.__kCoaxDiameter * self.__kNumberCoax/self.__kStackCoax)  # size of box containing coax. n by n stacking
        self.__fCoax_Y   = self.Round(self.__kCoaxDiameter * self.__kStackCoax)  # size of box containing coax. n by n stacking
        #
        # A tube has to fit around the maximum dimension of this stack. 
        # 
        self.__fD_Inner  = np.sqrt(np.power(self.__fCoax_X,2.0) + np.power(self.__fCoax_Y,2.0))
        self.__fD_100mk  = self.Round(self.__fD_Inner + 2.0 * self.__kRStep)
        self.__fD_800mk  = self.Round(self.__fD_100mk + 2.0 * self.__kRStep)
        self.__fD_4k     = self.Round(self.__fD_800mk + 2.0 * (self.__kRStep + self.__kInnerShieldThickness))
        self.__fD_50k    = self.Round(self.__fD_4k    + 2.0 * (self.__kRStep + self.__kInnerShieldThickness))
        self.__fD_300k   = self.Round(self.__fD_50k   + 2.0 * (self.__kRStep + self.__k50kShieldThickness))
       
    def CalculateQDot(self):
        """
        @brief - populate QDot for each stage. 
        Heat loss by radiative processes only at this point. 
        """
        self.__QDot_Inner[0] = RadiativeCylinder(self.__fD_Inner/2.0, self.__fD_800mk/2.0, self.__kE_Gold, self.__kE_Gold, self.__kLength, self.__kT10mk, self.__kT800mk)
        self.__QDot_800mk[0] = RadiativeCylinder(self.__fD_800mk/2.0, self.__fD_4k/2.0, self.__kE_Gold, self.__kE_Gold, self.__kLength, self.__kT800mk, self.__kT4k)
        self.__QDot_4k[0]    = RadiativeCylinder(self.__fD_4k/2.0, self.__fD_50k/2.0, self.__kE_Gold, self.__kE_Al, self.__kLength, self.__kT4k, self.__kT50k)
        self.__QDot_50k[0]   = RadiativeCylinder(self.__fD_50k/2.0, self.__fD_300k/2.0, self.__kE_Al, self.__kE_Al, self.__kLength, self.__kT50k, self.__kT300k)
        #print("DEBUG")
        #print(' DEBUG: ',self.__fD_Inner/2.0, " ", self.__fD_800mk/2.0, " ", self.__kE_Gold, " ", self.__kE_Gold, " ", self.__kLength, " ", self.__kT10mk, " ", self.__kT800mk)

        
        # conductivity through support elements, use specified thermal conductivity
        A = 0.01 * 0.01 * 4    # QTY: 4, 1 x 1 cm support.

        dT = self.__kT800mk - self.__kT10mk
        l  = self.__fD_800mk/2.0 - self.__fD_Inner/2.0
        self.__QDot_Inner[1] = Conduction(self.__kK_Insulator, l, A, dT)
        print('DEBUG: ', self.__QDot_Inner[1], ' ', self.__kK_Insulator, ' L: ', l,' ', A, ' ', dT)

        dT = self.__kT4k - self.__kT800mk
        l =  self.__fD_4k/2.0 - self.__fD_800mk/2.0
        self.__QDot_800mk[1] = Conduction(self.__kK_Insulator, l, A, dT)
              
        dT = self.__kT50k - self.__kT4k
        l =  self.__fD_50k/2.0 - self.__fD_4k/2.0
        self.__QDot_4k[1] = Conduction(self.__kK_Insulator, l, A, dT)

        dT = self.__kT300k - self.__kT50k
        l =  self.__fD_300k/2.0 - self.__fD_50k/2.0
        self.__QDot_50k[1] = Conduction(self.__kK_Insulator, l, A, dT)

    
    def __str__(self):
        """
        @brief - allow the user to easily output the parameters to the screen. 
        """
        output = " ---------------------------------------------- \n"
        output += " Coax details: \n"
        output += "     Coax Diameter: " + str(self.__kCoaxDiameter) + "\n"
        output += "       Number coax: " + str(self.__kNumberCoax) + "\n"
        output += "          Stacking: " + str(self.__kStackCoax) + "\n"
        output += " for a box size of: " + str(self.__fCoax_X) + " " + str(self.__fCoax_Y) + "\n"
        output += "\n"
        output += "Final Shield diameters, inner Diameter dimensions. \n"
        output += "  gold plated copper 800mK: " + str(self.__fD_800mk) + "\n"
        output += "     gold plated copper 4K: " + str(self.__fD_4k)  + "\n"
        output += "    gold plated copper 50k: " + str(self.__fD_50k)  + "\n"
        output += "             aluminum 300k: " + str(self.__fD_300k) + "\n"
        output += "\n"
        output += "             Tunnel Length: " + '{:6.2f}'.format(self.__kLength) + "\n"
        output += " ============= \n"
        output += " Radiative Heat Loads             Conductive Heat Loads using " + self.__kK_InsulatorName + ":\n"
        output += "                  onto 50K: " + '{:6.2g}'.format(self.__QDot_50k[0])   + " " + '{:6.2g}'.format(self.__QDot_50k[1])   + " Watts \n"
        output += "                   onto 4K: " + '{:6.2g}'.format(self.__QDot_4k[0])    + " " + '{:6.2g}'.format(self.__QDot_4k[1])    + " Watts \n"
        output += "                onto 800mk: " + '{:6.2g}'.format(self.__QDot_800mk[0]) + " " + '{:6.2g}'.format(self.__QDot_800mk[1]) + " Watts \n"
        output += "                 onto coax: " + '{:6.2g}'.format(self.__QDot_Inner[0]) + " " + '{:6.2g}'.format(self.__QDot_Inner[1]) + " Watts \n"
        output += " -----------------------------------------------\n"
        return output
        
    def Draw(self):
        """
        @brief - Make an end on view of the tunnel. 
        """
        circle1 = plt.Circle((0.5,0.5),self.__fD_800mk/2.0, color='r', fill=False)
        circle2 = plt.Circle((0.5,0.5),self.__fD_4k/2.0, color='g', fill=False)
        circle3 = plt.Circle((0.5,0.5),self.__fD_50k/2.0, color='b', fill=False)
        circle4 = plt.Circle((0.5,0.5),self.__fD_300k/2.0, fill=False)

        xdim = self.__fD_Inner
        x0   = 0.5 - self.__fCoax_X/2.0
        y0   = 0.5 - self.__fCoax_Y/2.0
        rect    = plt.Rectangle((x0, x0), self.__fCoax_X, self.__fCoax_Y, color='y',fill=True)
        
        fig, ax = plt.subplots()
        ax.add_patch(circle1)
        ax.add_patch(circle2)
        ax.add_patch(circle3)
        ax.add_patch(circle4)
        ax.add_patch(rect)
        
        plt.grid(True)
```

```python
def QDotCyl(Radius, T1, T2):
    """!return QDot and T range based on input parameters:
    @param Radius of outer cylinder in meters
    @param T1 - temperature of outer cylinder in K
    @param T2 - temperature of inner cylinder in K
    @return QDot (J/s or watts) and temperature range 
    """
    step = (T1-T2)/100
    T = np.arange(T2, T1, step)
    Area = CylinderArea(Radius, 1.0)
    Qdot = Radative(1.0, Area, T2, T)
    return [T, Qdot]
```
