```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

```python
# 1. Define the domain
x = np.linspace(-10, 10, 100)
y = np.linspace(-10, 10, 100)
X, Y = np.meshgrid(x, y)
Ri = 1.0        # Torus radius. 
# 2. Calculate the distance from the origin
R = np.sqrt(X**2 + Y**2)

# 3. Calculate the z-values (2D sinc function)
# Handle the case where R is zero to avoid division by zero
Z = np.where(R == 0, 1.0, np.sin(R) / R) 

# 4. Create the 3D plot
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Plot the surface
ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none')

# Set labels and title
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis (sinc(R))')
ax.set_title('3D Sinc Function')

# 5. Display the plot
plt.show()
```
