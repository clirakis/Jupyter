<H1> Equations </H1>
Transform Lat-Lon-Altitude to ECEF XYZ. <br> 
Latitude $\phi$ <br> 
Longitude $\lambda$ <br> 
ellipsoidal height h <br> 
WGS  constants <br> 
$\alpha = 6378137.0$ meters <br> 
$f = {1.0\over{298.257223563}} $ <br>
$e^2 = f(2-f)$ <br> 
Equations<br> 
$N(\phi) = {\alpha \over {\sqrt{1-e^2 sin^2(\phi)}}}$<br> 
$X = (N+h)cos(\phi)cos(\lambda)$ <br> 
$Y = (N+h)cos(\phi)sin(\lambda)$ <br>
$Z = [N(1-e^2 )+h ] sin(\phi)$<br>

```python
def ECEF(Lat, Lon, Alt):
    """
    Lat in degrees
    Lon in degrees
    Alt in meters based on ellipsoid
    """
    alpha = 6378137.0
    f = 1.0/298.257223563
    e2 = f*(2-f)
    # convert Lat and Lon to radians. 
    Phi = Lat * np.pi/180.0
    Lambda = Lon * np.pi/180.0
    N = alpha/np.sqrt(1.0-e2*np.power(np.sin(Phi),2)) # error was here
    # actual transform
    # Alt should be reference to the height above the WGS84 ellipsoid
    X = (N+Alt)*np.cos(Phi)*np.cos(Lambda)
    Y = (N+Alt)*np.cos(Phi)*np.sin(Lambda)
    Z = (N*(1-e2)+Alt)*np.sin(Phi)

    return X,Y,Z
```

```python
# test case 1
X,Y,Z = ECEF(0.0, 0.0, 0.0)
print("My ECEF - X: ", X, " Y= ", Y, " Z= ", Z)
```
