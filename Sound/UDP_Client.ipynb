<a href="https://python-osc.readthedocs.io/en/latest/client.html#module-pythonosc.udp_client"> example UDP client</a>

```python
# My imports
import numpy as np
import matplotlib.pyplot as plt
import time
from pythonosc.udp_client import SimpleUDPClient
```

```python
Address = "127.0.0.1"
port = 5005
#
client = SimpleUDPClient(Address,port)
client.send_message("/some/address", 1234)
client.send_message("/some/address",[1,2.,"hello"])
```

```python
# Alternatively, use a context manager to automatically close the socket
with SimpleUDPClient(Address, port) as client:
    client.send_message("/some/address", 123)
```

```python
Address = "127.0.0.1"
port = 5005
#
client = SimpleUDPClient(Address,port)
client.send_message("/text", "My Text")
client.send_message("/number", 3.14)
```

<H1>Oscillator </H1> 
/osc/number frequency, phase <br>
<H1> ADSR</H1>
/adsr/number A,D,S,R <br> 
Try these out. 

```python
Address = "127.0.0.1"
port = 5005
#
client = SimpleUDPClient(Address,port)
client.send_message("/osc/1", [440.0,0.0 ])
time.sleep(4)
client.send_message("/osc/1", [220.0,0.0 ])
```
