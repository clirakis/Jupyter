```python
# My imports

import numpy as np
import matplotlib.pyplot as plt
```

```python
# Generated by chatGPT
# 64 evenly spaced grayscale values from black to white
#
values = np.linspace(0, 255, 64).round().astype(int)
grid   = values.reshape(8,8)
#
fig, ax = plt.subplots(figsize=(8,8))
#
ax.imshow(
    grid, 
    cmap='gray',
    vmin=0,
    vmax=255,
    interpolation="nearest"
)

# add the digital value to each square. 
for row in range(8):
    for col in range(8):
        value = grid[ row, col]

        # switch label color for readabiltiy
        if (value < 128):
            text_color = "white"  
        else: 
            text_color = "black"

            
        ax.text(
            col, 
            row,
            str(value),
            ha="center",
            va="center",
            color=text_color,
            fontsize=9,
        )

# grid lines between patches. 
ax.set_xticks(np.arange(-0.5, 8, 1), minor=True)
ax.set_yticks(np.arange(-0.5, 8, 1), minor=True)
ax.grid(which="minor", linewidth=0.5)
#
ax.tick_params(
    which="both",
    bottom=False,
    left=False,
    labelbottom=False,
    labelleft=False, 
)

ax.set_title("Cyanotype Grayscale Test - 64 levels", pad=15)

plt.tight_layout()
plt.savefig("cyanotype_grayscale.png", dpi=600, bbox_inches="tight")
plt.show()
```

When printing this, it turns out the "print side" of the transparency stuff smears more but absorbs the ink better. The non-print side is more clearly defined.
