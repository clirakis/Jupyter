```python
# imports - package libraries etc. May be overkill
import numpy as np
import matplotlib.pyplot as plt
import skrf as rf
from skrf import Network
from skrf import Frequency
from skrf.circuit import Circuit
from skrf.tlineFunctions import INF
import scipy.constants as konst
#
# skrf figure styling
rf.stylely()
```

```python
def BigLoop(Frequency):
    """
    Calculate the parameters of the big loop. 
    @param Frequency - center frequency in HZ of antenna
    """
    Lambda = konst.c/Frequency      # wavelength 
    Height   = 0.260 * Lambda
    Diameter = 0.173 * Lambda
    LegSize  = 0.560 * Lambda
    return Height, Diameter, LegSize
```

```python
def SmallLoop(Frequency):
    """
    Calculate the parameters of the small loop. 
    @param Frequency - center frequency in HZ of antenna
    """
    Lambda = konst.c/Frequency      # wavelength 
    Height   = 0.238 * Lambda
    Diameter = 0.156 * Lambda
    LegSize  = 0.508 * Lambda
    return Height, Diameter, LegSize
```

```python
#F = 435.0e6
F = 436.0e6
[Height, Diameter, LegSize] = BigLoop(F)
print("Big Loop, Height = ", Height, " Diameter = ", Diameter, " LegSize = ", LegSize, " meters.")
[Height, Diameter, LegSize] = SmallLoop(F)
print("Sma Loop, Height = ", Height, " Diameter = ", Diameter, " LegSize = ", LegSize, " meters.")
```

```python
#
# Lets look at the sensitivity of the dimensions from frequency. 
#
F0 = np.arange(434.0, 438.0, 0.1)
F0 = F0 * 1.0e6
[Height, Diameter, LegSize] = BigLoop(F0)
plt.plot(F0, LegSize/0.0254)
```
