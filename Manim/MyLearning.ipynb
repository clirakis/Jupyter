```python
class CBL1(Scene):
    def construct(self):
        """
        I think of this as the static background when called, setup all the base pieces. 
        I started with the Manim logo class and am fooling with that first. 
        """
        # Set background color
        self.camera.background_color = "#ece6e2"
        # colors for objects. 
        # the pound sign indicates hex, Two byte objects for RGB (0:255)
        logo_green = "#87c2a5"
        logo_blue = "#525893"
        logo_red = "#e07a5f"
        logo_black = "#343434"
        # create a text box, set the color and scale it. 
        ds_m = MathTex(r"\mathbb{CBL}", fill_color=logo_black).scale(7)
        # create a translation to the text. 
        ds_m.shift(2.25 * LEFT + 1.5 * UP)
        # mine, put a star at the center of the diagram. 
        star = Star(n=5, outer_radius=2.0, color=YELLOW, fill_color=YELLOW_E, fill_opacity=0.5)
        #
        circle = Circle(color=logo_green, fill_opacity=1).shift(LEFT)
        square = Square(color=logo_blue, fill_opacity=1).shift(UP)
        triangle = Triangle(color=logo_red, fill_opacity=1).shift(RIGHT)
        logo = VGroup(star, triangle, square, circle, ds_m)  # order matters
        logo.move_to(ORIGIN)
        self.add(logo)
```

```python
manim -qm -v WARNING CBL1
```
