```python
# My imports
import numpy as np
import matplotlib.pyplot as plt
from manim import *
```

Animate a circle

```python
class TicTacToe(Scene):
    def construct(self):
        # 1. Create the Board Grid lines
        line_kwargs = {"color": BLUE, "stroke_width": 6}
        
        # Horizontal lines
        h1 = Line(start=[-3, 1, 0], end=[3, 1, 0], **line_kwargs)
        h2 = Line(start=[-3, -1, 0], end=[3, -1, 0], **line_kwargs)
        
        # Vertical lines
        v1 = Line(start=[-1, 3, 0], end=[-1, -3, 0], **line_kwargs)
        v2 = Line(start=[1, 3, 0], end=[1, -3, 0], **line_kwargs)
        
        board = VGroup(h1, h2, v1, v2)
        
        # 2. Define coordinate helpers for the 3x3 grid spaces
        # Map grid positions (row, col) from 0 to 2 to screen coordinates
        positions = {
            (0, 0): [-2, 2, 0],  (0, 1): [0, 2, 0],  (0, 2): [2, 2, 0],
            (1, 0): [-2, 0, 0],  (1, 1): [0, 0, 0],  (1, 2): [2, 0, 0],
            (2, 0): [-2, -2, 0], (2, 1): [0, -2, 0], (2, 2): [2, -2, 0],
        }

        # 3. Helper functions to generate X and O Mobjects
        def make_x(pos):
            cross = Cross(scale_factor=0.6, stroke_width=8, stroke_color=RED)
            cross.move_to(positions[pos])
            return cross

        def make_o(pos):
            circle = Circle(radius=0.5, stroke_width=8, color=GREEN)
            circle.move_to(positions[pos])
            return circle

        # --- ANIMATION SEQUENCE ---
        
        # Draw the board
        self.play(Create(board))
        self.wait(0.5)

        # Move 1: X takes center (1, 1)
        self.play(Create(make_x((1, 1))), run_time=0.7)
        self.wait(0.3)

        # Move 2: O takes top left (0, 0)
        self.play(Create(make_o((0, 0))), run_time=0.7)
        self.wait(0.3)

        # Move 3: X takes top center (0, 1)
        self.play(Create(make_x((0, 1))), run_time=0.7)
        self.wait(0.3)

        # Move 4: O takes bottom center (2, 1)
        self.play(Create(make_o((2, 1))), run_time=0.7)
        self.wait(0.3)

        # Move 5: X takes top right (0, 2) to win!
        x_winning_move = make_x((0, 2))
        self.play(Create(x_winning_move), run_time=0.7)
        self.wait(0.5)

        # 4. Highlight the winning strike row
        winning_line = Line(start=[-3, 2, 0], end=[3, 2, 0], color=YELLOW, stroke_width=10)
        win_text = Text("X Wins!", color=RED).scale(1.5).to_edge(UP)
        
        self.play(Create(winning_line), Write(win_text))
        self.wait(2)
```
